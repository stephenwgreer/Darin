```python
### API call to Claude to analyze a txt file and pull out relevant topics

import anthropic
import os

from dotenv import load_dotenv

load_dotenv()

# Read the text file
with open("transcription.txt", "r") as file:
    text_content = file.read()

PROMPT = f"""
Extract the top 3 main topics from the following text. For each topic:
1. Provide a concise label (1-3 words)
2. Include 2-3 supporting keywords or phrases from the text
3. Assign a relevance score (1-10)

Text to analyze:
{text_content}
"""
```

```python
client = anthropic.Anthropic(
    # defaults to os.environ.get("ANTHROPIC_API_KEY")
    api_key=os.getenv("ANTHROPIC_API_KEY"),
)
message = client.messages.create(
    model="claude-3-7-sonnet-20250219",
    max_tokens=1024,
    messages=[
        {"role": "user", "content": PROMPT}
    ]
)

response=message.content
```
