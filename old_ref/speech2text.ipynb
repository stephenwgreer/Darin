```python
import io
import os
import requests
import base64
import json
from dotenv import load_dotenv

load_dotenv()

def speech_to_text_with_api_key(wav_file_path="recording.wav", api_key=None):
    """
    Use Google Cloud Speech-to-Text API with an API key instead of service account
    """
    if not api_key:
        raise ValueError("API key is required")
    
    # Read the audio file
    with open(wav_file_path, "rb") as audio_file:
        content = audio_file.read()
    
    # Encode audio content as base64
    audio_content = base64.b64encode(content).decode('utf-8')
    
    # Prepare the request JSON
    request_data = {
        "config": {
            "encoding": "LINEAR16",
            "sampleRateHertz": 48000,  # Match your audio's sample rate
            "languageCode": "en-US",
            "enableAutomaticPunctuation": True
        },
        "audio": {
            "content": audio_content
        }
    }
    
    # API endpoint
    url = f"https://speech.googleapis.com/v1/speech:recognize?key={api_key}"
    
    # Make the request
    try:
        response = requests.post(url, json=request_data)
        response.raise_for_status()  # Raise exception for HTTP errors
        
        # Parse the response
        result = response.json()
        
        # Extract and return the transcribed text
        text = ""
        if "results" in result:
            for res in result["results"]:
                text += res["alternatives"][0]["transcript"]
        
        print("Transcription: ", text)
        return text
        
    except Exception as e:
        print(f"Error with Google Cloud Speech API: {e}")
        return ""

# Example usage:
API_KEY = os.getenv("GOOGLE_API_KEY")
result = speech_to_text_with_api_key("recording.wav", api_key=API_KEY)
print(f"Final result: {result}")

## write result to file
with open("transcription.txt", "w") as text_file:
    text_file.write(result)
```

```python
# Install required packages
# pip install python-dotenv requests

import os
import requests
import json
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Get the API key from environment variables
DEEPGRAM_API_KEY = os.getenv("DEEPGRAM_API_KEY")

# Ensure the API key is available
if not DEEPGRAM_API_KEY:
    raise ValueError("DEEPGRAM_API_KEY not found in .env file")

def transcribe_audio(audio_file_path):
    """
    Transcribe an audio file using Deepgram's API
    
    Args:
        audio_file_path (str): Path to the audio file
    
    Returns:
        dict: Transcription results from Deepgram
    """
    # Deepgram API endpoint
    url = "https://api.deepgram.com/v1/listen"
    
    # Request headers
    headers = {
        "Authorization": f"Token {DEEPGRAM_API_KEY}"
    }
    
    # Parameters for the transcription
    params = {
        "punctuate": "true",
        "model": "general",
        "language": "en-US"
    }
    
    # Open the audio file
    with open(audio_file_path, "rb") as audio:
        # Send the request to Deepgram
        response = requests.post(
            url,
            headers=headers,
            params=params,
            data=audio
        )
    
    # Check if the request was successful
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code}")
        print(response.text)
        return None

# Example usage
if __name__ == "__main__":
    # Replace with your audio file path
    audio_path = "BSGPT_REC.wav"
    
    # Get the transcription
    result = transcribe_audio(audio_path)
    
    if result:
        # Extract the transcript
        transcript = result["results"]["channels"][0]["alternatives"][0]["transcript"]
        print("Transcription:")
        print(transcript)
```
